##  include csv file

```python
Performance=pd.read_csv(filepath_or_buffer="StudentsPerformance.csv")
```

## Performance

```python
Performance.describe()
```

## describe the performance

```python
Performance.describe(include="all")
```

## finding the students average mean using the parents education level

```python
Performance.groupby(by="parental level of education")["writing score"].mean().sort_values()
```

## finding all kind of parental level education

```python
Performance["parental level of education"].unique()
```

## finding the  number of unique parental level

```python
Performance["parental level of education"].nunique()
```

```python
Performance.isnull()
```

## Calculate correlation matrix between the three scores

```python

Performance[['math score', 'reading score', 'writing score']].corr()
```

## Compare average scores between completed vs. none preparation groups

```python

Performance.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()
```

## Average scores segmented by lunch type (standard vs. free/reduced)

```python

Performance.groupby('lunch')[['math score', 'reading score', 'writing score']].mean()

```

## Average scores segmented by gender



```python
Performance.groupby('gender')[['math score', 'reading score', 'writing score']].mean()
```
